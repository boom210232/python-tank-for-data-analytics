```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
## Display all the columns of the dataframe
pd.pandas.set_option('display.max_columns',None)
```

```python
dataset=pd.read_csv('train.csv')
## print shape of dataset with nb-rows and nb-columns
```

```python
## print the top5 records
top5_rc = dataset.head()
top5_rc
```

## Missing Values

```python
## Use heatmap tp show missing values features
sns.heatmap(dataset.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
## Here we will check the percentage of nan values present in each feature
## 1 -step make the list of features which has missing values
## 2- step print the feature name and the percentage of missing values
null_percent = dataset.isnull().sum()/len(dataset)
missing_data_more_zero = null_percent[null_percent > 0]

for feat, percent in zip(missing_data_more_zero.index, missing_data_more_zero.values):
    print(f"{feat}: {percent:.4f} % missing values")
```

#### Temporal Variables(Eg: Datetime Variables)

```python
# list of variables that contain year information, year variables
year_var = [column for column in dataset.columns if "YR" in column.upper() or "YEAR" in column.upper() ]
year_var
```

```python
# let's explore the content of these year variables
y_built = dataset["YearBuilt"].unique()
y_remodAdd = dataset["YearRemodAdd"].unique()
y_gar_built = dataset["GarageYrBlt"].unique()
y_sold = dataset["YrSold"].unique()

# For show
print(f"YearBuilt {y_built}")
print(f"YearRemodAdd {y_remodAdd}")
print(f"GarageYrBlt {y_gar_built}")
print(f"YrSold {y_sold}")
```

```python
## Lets analyze the Temporal Datetime Variables
## We will check whether there is a relation between year the house is sold and the sales price
# by_year_sale = dataset.groupby("YrSold")["SalePrice"].apply(lambda x: np.average(x.dropna()))
by_year_sale = dataset.groupby('YrSold')['SalePrice'].mean()
plt.figure(figsize=(10, 6))
plt.title("House Price vs YearSold")
plt.xlabel("Year Sold")
plt.ylabel("Median House Price")
plt.plot(by_year_sale.index,by_year_sale.values)


```

```python
## Numerical variables are usually of 2 type, Continous variable and Discrete Variables (finite set of values < 25)
numerical_var = dataset.select_dtypes(include=['int64', 'float64']).columns
discrete_feature = [column for column in numerical_var if dataset[column].nunique() < 25]
print(f"Discrete Variables Count: {len(discrete_feature)}")
```

```python
discrete_feature
```

```python
dataset[discrete_feature].head()
```

```python
## Lets Find the realtionship between Discrete Variables and Sale PRice
fig ,axes= plt.subplots(len(discrete_feature), 1, figsize=(10, 4 * len(discrete_feature)))
## plt.subplots == plt.subplot in many many time at once ต้องมาคู่กันทั้ง  fig , axes

axes = axes.flatten()
for i, j in enumerate(discrete_feature):
    avg_sale_price = dataset.groupby(j)["SalePrice"].mean()
    avg_sale_price.plot(kind = "bar", ax=axes[i])
    # dataset[j].hist( ax=axes[i], )
    axes[i].set_title(f'{j}')
    axes[i].set_xlabel(j)
    axes[i].set_ylabel('SalePrice')
plt.tight_layout()
plt.show()
```

#### Continuous Variable

```python
numerical_var = dataset.select_dtypes(include=['int64', 'float64']).columns
cons_feature = [column for column in numerical_var if dataset[column].nunique() > 25]
print(f"Continuous feature Count: {len(cons_feature)}")
```

```python
## Lets analyse the continuous values by creating histograms to understand the distribution
fig ,axes= plt.subplots(len(cons_feature)-1, 1, figsize=(8, 4 * len(cons_feature)-1))
## plt.subplots == plt.subplot in many many time at once ต้องมาคู่กันทั้ง  fig , axes

if len(cons_feature) == 1:
    axes = [axes]

axes = axes.flatten()
for i, j in enumerate(cons_feature[1:],start=1):

    # avg_sale_price = dataset.groupby(j)["SalePrice"].mean()
    # avg_sale_price.plot(kind = "bar", ax=axes[i])
    dataset[j].hist( ax=axes[i-1], bins=40)
    axes[i-1].set_title(f'{j}')
    axes[i-1].set_xlabel(j)
    axes[i-1].set_ylabel("Count")
plt.tight_layout()
plt.show()
```
