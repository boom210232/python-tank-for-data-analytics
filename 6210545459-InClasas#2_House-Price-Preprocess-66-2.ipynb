## Housing Prices- Pre-processing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%matplotlib inline
# to visualise al the columns in the dataframe
pd.pandas.set_option('display.max_columns', None)
```

```python
dataset=pd.read_csv('train.csv')
dataset.head()
```

```python
dataset.columns
```

## Missing Values

```python
sns.heatmap(dataset.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
## List all categorical features which are missing
categorical_columns = dataset.select_dtypes(include=['object']).columns

for i in categorical_columns:
    missing_percent = (dataset[i].isnull().sum() / len(dataset))
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.4f}% missing values")
```

```python
## Replace missing value categorical features with a new label "Missing"
categorical_columns = dataset.select_dtypes(include=['object']).columns
for i in categorical_columns:
    dataset[i].fillna("Missing", inplace=True)

# ถ้าแก้หมดต้องไม่มีอะไรออกมา
for i in categorical_columns:
    missing_percent = (dataset[i].isnull().sum() / len(dataset))
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.4f}% missing values")
```

```python
dataset.head()
```

```python
## Now lets check for numerical variables the contains missing values
numerical_columns = dataset.select_dtypes(exclude=['object']).columns

for i in numerical_columns:
    missing_percent = (dataset[i].isnull().sum() / len(dataset))
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.4f}% missing values")
```

```python
## Replacing the numerical Missing Values using Median
for i in numerical_columns:
    median_v = dataset[i].median()
    dataset[i].fillna(median_v, inplace=True)

# ถ้าแก้หมดต้องไม่มีอะไรออกมา
for i in numerical_columns:
    missing_percent = (dataset[i].isnull().sum() / len(dataset))
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.4f}% missing values")
```

```python
## Temporal Variables (Date Time Variables)

for feature in ['YearBuilt','YearRemodAdd','GarageYrBlt']:

    dataset[feature]=dataset['YrSold']-dataset[feature]
```

## Numerical Variables
Since the numerical variables are skewed we will perform log normal distribution

```python
# Take log

numerical_columns = dataset.select_dtypes(exclude=['object']).columns[1:]
for i in numerical_columns:
    if (dataset[i] <= 0).any():
        continue
    dataset[i] = np.log(dataset[i])
```
