```python
arr_keep = 0

for key,item in d.items():
  if key == "text":
    # print(item)
    for i in range (len(item) - 4):
      if (len(item[i])==1) and (len(item[i+1])==4) and (len(item[i+2])==5) and (len(item[i+3])==2) and (len(item[i+4])==1):
        print("Thai national ID card number is:")
        print(item[i] + " " + item[i+1] + " " + item[i+2] + " " + item[i+3] + " " + item[i+4] + " ")
        arr_keep = i

print("Find ID at array: " + str(arr_keep))
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image



plt.rcParams['figure.figsize'] = [10, 10]

top = 0
left = 0
bottom = 0
right = 0
width = 0
end_width = 0

url = 'https://www.rd-comp.com/wp-content/uploads/2018/10/AThaiID_P4.jpg'
response = requests.get(url)
img = Image.open(BytesIO(response.content))
plt.figure(figsize = (20, 10))

# Create figure and axes
fig, ax = plt.subplots()

# Display the image
ax.imshow(img)

for key,item in d.items():
  if key == "top":
    # print(item)
    top = item[arr_keep]
  if key == "height":
    # print(item)
    bottom = item[arr_keep] 
  if key == "left":
    # print(item)
    left = item[arr_keep]
    right = item[arr_keep+4]
  if key == "width":
    # print(item)
    width = item[9]
    end_width = item[arr_keep+4]


# Create a Rectangle patch
rect = patches.Rectangle((left,top ), right-left+end_width, (bottom), linewidth=2, edgecolor='r', facecolor='none')

# Add the patch to the Axes
ax.add_patch(rect)

# plt.imshow(img, interpolation='nearest')
plt.show()
```
