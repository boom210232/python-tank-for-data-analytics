## 5) COVID cases comparison #2 (3 points)

Plot the same data above, but convert the data in y axis from `number of cases` to `percentage of its population`. Use population data provided by this file: https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/jhu/locations.csv

```python
# Your code here
population = pd.read_csv('https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/jhu/locations.csv')
population
```

```python
covid2 = covid.copy()
for covids in range(len(covid)):
  for popu in range(len(population)):
    if covid2['location'][covids] == population["location"][popu]:
      covid2['total_cases'][covids] = (covid2['total_cases'][covids] / population["population"][popu])*100

covid2
# เอาแค่conceptก่อนนะครับ runtime รอไม่ไหว
```
