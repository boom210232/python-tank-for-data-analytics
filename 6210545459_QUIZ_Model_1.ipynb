### Q1 Show number of good students? Good students are those with GPA greater or  equal to 3.0.

### Q2 Show number of good students who have already graduated? Good students are  those, which graduated with GPA greater or equal to 3.0.

```python
good_grad = (dataframe_join['STATUS'] == 'G') & (dataframe_join['GPAX'] >= 3.0)
good_grad_c = good_grad.sum()
print(good_grad_c)
```

### Q3 Show number of bad students who have already graduated? Bad students are  those, which graduated with GPA less than 3.0.

```python
bad_grad = (dataframe_join['STATUS'] == 'G') & (dataframe_join['GPAX'] < 3.0)
bad_grad_c = bad_grad.sum()
print(bad_grad_c)
```

# Preprocessing

### Remove missing data

```python
import seaborn as sns
import seaborn as sn
```

```python
sns.heatmap(dataframe_join.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
data_clean = dataframe_join.copy()
data_clean.drop(columns='GPA', inplace=True)
```

```python
sns.heatmap(data_clean.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
## List all categorical features which are missing
categorical_columns = data_clean.select_dtypes(include=['object']).columns
print(categorical_columns)

for i in categorical_columns:
    missing_percent = (data_clean[i].isnull().sum() / len(data_clean))*100
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.8f}% missing values")
```

```python
## List all numerical features which are missing
numerical_columns = data_clean.select_dtypes(exclude=['object']).columns
print(numerical_columns)

for i in numerical_columns:
    missing_percent = (data_clean[i].isnull().sum() / len(data_clean))*100
    if missing_percent > 0:
        print(f"{i}: {missing_percent:.8f}% missing values")
```

# Model

```python
data_clean
```

```python
data_clean.dtypes
```

```python
input = data_clean.copy()
input.drop(columns=['STUDENT_ID', 'AGE_1_YEAR', 'ZIP_CODE', 'DEPARTMENT_ID', 'YEAR_ADMIT','SEMESTER_ADMIT'], inplace=True)
input.drop(columns=['TITLE_NAME', 'FIRST_NAME', 'LAST_NAME', 'GENDER', 'STATUS'], inplace=True)
y=data_clean['DEPARTMENT_ID']
```

```python
input.dtypes
```

```python
input.head(5)
```

### Splitting the dataset into the Training set and Test set

```python
X = input.values
print(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
print(X_train)
```

```python
print(y_train)
```

### Training the Decision Tree Classification model on the Training set

```python
from sklearn import tree
classifier = tree.DecisionTreeClassifier(max_depth = 3, criterion = 'gini', random_state = 0)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
```

```python
plt.figure(figsize=(15,10))
tree.plot_tree(classifier,filled=True)
```

```python
print(tree.export_text(classifier))
```

### Model Evaluation

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
```

### Making the Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```
