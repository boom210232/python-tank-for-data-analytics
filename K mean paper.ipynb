```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Define the data
data = {'X': [3, 9, 4, 2, 10, 2, 6, 4, 4],
        'Y': [8, 4, 9, 2, 5, 4, 8, 3, 7]}

df = pd.DataFrame(data)

# Perform k-means clustering with k=3
kmeans = KMeans(n_clusters=3)
kmeans.fit(df)

# Add cluster labels to the DataFrame
df['Cluster'] = kmeans.labels_

# Extract centroids
centroids = pd.DataFrame(kmeans.cluster_centers_, columns=['X', 'Y'])

# Plot the data points with different colors for each cluster
colors = ['red', 'green', 'blue']
for cluster, color in zip(range(3), colors):
    plt.scatter(df[df['Cluster'] == cluster]['X'], df[df['Cluster'] == cluster]['Y'], c=color, label=f'Cluster {cluster + 1}')

# Plot centroids
plt.scatter(centroids['X'], centroids['Y'], c='black', marker='X', s=100, label='Centroids')

# Add labels and legend
plt.xlabel('X')
plt.ylabel('Y')
plt.title('K-Means Clustering with k=3')
plt.legend()

# Show the plot
plt.show()

# Print array numbers for each cluster
print("Array Numbers in Each Cluster:")
for cluster in range(3):
    array_numbers = df[df['Cluster'] == cluster].index + 1
    print(f'Cluster {cluster + 1}: {array_numbers.tolist()}')
```
